```python
import os
import psycopg2
import logging

# Set up logging
logging.basicConfig(filename='detection_process.log', level=logging.INFO, format='%(asctime)s - %(message)s')

# PostgreSQL Connection Details
PG_HOST = "localhost"  # PostgreSQL host
PG_USER = "postgres"   # Default PostgreSQL username
PG_PASSWORD = "1240"   # Your password
PG_DATABASE = "telegram_raw_data"  # Your database name

# Connect to PostgreSQL database
try:
    conn = psycopg2.connect(
        host=PG_HOST,
        user=PG_USER,
        password=PG_PASSWORD,
        dbname=PG_DATABASE
    )
    cursor = conn.cursor()
    logging.info("Connected to PostgreSQL database successfully.")
    print("Connected to PostgreSQL database successfully.")
except psycopg2.Error as e:
    logging.error(f"Error connecting to PostgreSQL: {str(e)}")
    print(f"Error connecting to PostgreSQL: {str(e)}")
    exit()
```

```python
# Create table if it doesn't exist
cursor.execute('''CREATE TABLE IF NOT EXISTS yolo_detections (
                    id SERIAL PRIMARY KEY,
                    image_name VARCHAR(255),
                    class_id INT,
                    bbox_x_center FLOAT,
                    bbox_y_center FLOAT,
                    bbox_width FLOAT,
                    bbox_height FLOAT,
                    confidence FLOAT
                 )''')

conn.commit()

# Paths
results_dir = "C:/Users/hp/kifiya_acadamy_week7/kifiya_week7_assignment/notebooks/yolov5/results/run13/labels"  # YOLOv5 detection results folder

# Loop through YOLO result text files
for txt_file in os.listdir(results_dir):
    if txt_file.endswith(".txt"):
        image_name = txt_file.replace(".txt", ".jpg")  # Assuming images are .jpg
        txt_path = os.path.join(results_dir, txt_file)

        try:
            with open(txt_path, "r") as file:
                lines = file.readlines()
                for line in lines:
                    data = line.strip().split()
                    if len(data) == 6:
                        class_id = int(data[0])
                        bbox_x_center = float(data[1])
                        bbox_y_center = float(data[2])
                        bbox_width = float(data[3])
                        bbox_height = float(data[4])
                        confidence = float(data[5])

                        # Insert data into PostgreSQL table yolo_detections
                        sql = """INSERT INTO yolo_detections 
                                 (image_name, class_id, bbox_x_center, bbox_y_center, bbox_width, bbox_height, confidence)
                                 VALUES (%s, %s, %s, %s, %s, %s, %s)"""
                        values = (image_name, class_id, bbox_x_center, bbox_y_center, bbox_width, bbox_height, confidence)

                        cursor.execute(sql, values)
                        conn.commit()
                        logging.info(f"Stored detection for {image_name}: Class {class_id}, BBox ({bbox_x_center}, {bbox_y_center}, {bbox_width}, {bbox_height}), Confidence {confidence}")

        except Exception as e:
            logging.error(f"Error processing {txt_file}: {str(e)}")

# Close database connection
cursor.close()
conn.close()
logging.info("Detection processing completed.")
```
